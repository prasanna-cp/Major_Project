```python
from ultralytics import YOLO
import cv2

# ---------------- ACCIDENT VIDEO DETECTION ----------------
def detect_accident_live(video_path, model_path, conf_thres=0.4):

    # Load YOLO model
    model = YOLO(model_path)

    # Open video
    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print("Error opening video")
        return

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        accident_detected = False

        # YOLO prediction on frame
        results = model(frame, conf=conf_thres, verbose=False)

        # Process detections
        for r in results:
            if r.boxes is not None:
                for box in r.boxes:

                    accident_detected = True

                    # Bounding box
                    x1, y1, x2, y2 = map(int, box.xyxy[0])

                    # Confidence
                    conf = float(box.conf[0])

                    # Draw box
                    cv2.rectangle(frame, (x1, y1), (x2, y2),
                                  (0, 255, 0), 2)

                    # Label
                    label = f"Accident {conf:.2f}"
                    cv2.putText(frame, label,
                                (x1, y1 - 10),
                                cv2.FONT_HERSHEY_SIMPLEX,
                                0.7, (0, 255, 0), 2)

        # Show alert text
        if accident_detected:
            cv2.putText(frame, "ACCIDENT DETECTED!",
                        (40, 50),
                        cv2.FONT_HERSHEY_SIMPLEX,
                        1.2,
                        (0, 0, 255),
                        3)

        # Display video
        cv2.imshow("Accident Detection - Live", frame)

        # Press Q to exit
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()
video_path = "test_video.mp4"
model_path = "runs/detect/train/weights/best.pt"

detect_accident_live(video_path, model_path)
```
